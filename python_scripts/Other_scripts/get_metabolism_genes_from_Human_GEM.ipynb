```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cobra
```

# Get genes from Human GEM model

```python
model = cobra.io.read_sbml_model("Human-GEM.xml")
```

```python
model
```

```python
genes = model.genes
genes = [gene.id for gene in genes]
genes = pd.DataFrame(genes, columns = ['gene_id'])
genes
```

#Add gene names

```python
gene_names = pd.read_csv('ensembl_id_gene_name.csv')
gene_names = gene_names.rename(columns = {'gene_name' : 'Symbol'})
gene_names
```

```python
gene_names = gene_names[gene_names['gene_id'].isin(genes.gene_id)]
gene_names = gene_names.set_index('gene_id')
gene_names
```

```python
res = pd.concat([gene_names, genes.set_index('gene_id')], axis = 'columns').dropna()
res
```

```python
res.to_csv('human_gem_associated_genes.csv', index = True)
```
