# Import modules

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

# Data

```python
rna = pd.read_csv('CCLE_OmicsExpressionProteinCodingGenesTPMLogp1BatchCorrected.csv', index_col = 0)
rna
```

```python
rna = rna.rename(columns=lambda x: x.split(' (')[0] if ' (' in x else x)
```

```python
rna = rna.loc[rna.index.notnull()]
rna
```

```python
cell_line_anotations = pd.read_csv('Cell_lines_annotations_20181226.txt', sep = '\t', usecols = ['CCLE_ID',	'depMapID',	'Name',	'Site_Primary'])
cell_line_anotations
```

```python
metabolites = pd.read_csv('CCLE_metabolomics_20190502.csv')
metabolites
```

```python
common_cell_lines = set(metabolites.DepMap_ID).intersection(set(rna.index))
common_cell_lines = list(common_cell_lines)
len(common_cell_lines)
```

```python
rna = rna.loc[common_cell_lines]
rna
```

```python
met_genes = pd.read_csv('human_gem_associated_genes.csv')
met_genes
```

```python
rna_met = rna[rna.columns[rna.columns.isin(met_genes['Symbol'])]]
rna_met
```

```python
rna_met.to_csv('CCLE_transcriptomics_tpm_metabolism_related_genes.csv', index = True)
```

```python
metabolites = metabolites[metabolites['DepMap_ID'].isin(common_cell_lines)]
metabolites
```

```python
metabolites = metabolites.set_index('DepMap_ID')
metabolites = metabolites.drop(columns = ['CCLE_ID'])
metabolites.index.name = None
metabolites
```

```python
metabolites.to_csv('CCLE_metabolites.csv', index = True)
```
