#Data

```python
data = pd.read_csv('drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/transcriptomics_tpm_benedetti.csv', index_col = 0)
data
```

```python
gene_names = pd.read_csv('drive/My Drive/Colab Notebooks/Deep_learning_metabolomics_prediction/BulkRNABert_data/benedetti_mart_export.txt', sep = '\t', index_col= 1)
gene_names.index.name = None
gene_names
```

```python
gene_names_dict = gene_names['Gene stable ID'].to_dict()
len(gene_names_dict)
```

```python
data['name'] = data.index.map(gene_names_dict)
data = data.dropna(subset = ['name'])
data = data.set_index('name')
data.index.name = None
data
```

```python
save_data = data.T
save_data.to_csv('drive/My Drive/Colab Notebooks/Deep_learning_metabolomics_prediction/gene_expression_benedetti_dataset.csv')
```

```python
with open("drive/My Drive/Colab Notebooks/Deep_learning_metabolomics_prediction/BulkRNABert_data/common_gene_id.txt") as f:
    gene_list = [line.strip() for line in f]

print(len(gene_list))
```

```python
genes_not_found_in_data = [gene for gene in gene_list if gene not in data.index]
len(genes_not_found_in_data)
```

```python
for i in genes_not_found_in_data:
  data.loc[i] = np.nan
data
```

```python
data = data.loc[gene_list]
data
```

```python
data = data.reindex(gene_list)
data
```

```python
#Assert gene order
good = []
for i in data.index:
  for j in gene_list:
    if i == j:
      good.append(i)
len(good)
```
