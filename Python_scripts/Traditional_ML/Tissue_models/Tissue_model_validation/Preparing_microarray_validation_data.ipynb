```python
samples_dict = sample_annot_microarrays[["MetabFile", "RNAFile"]].set_index("RNAFile").to_dict()["MetabFile"]
samples_dict
```

```python
met_genes = pd.read_csv('drive/My Drive/Colab Notebooks/GP_lab_data/FBA_metabolic_atlas_models/human_gem_associated_genes.csv')
met_genes
```

```python
import joblib

model = joblib.load('drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/Tissue_res/ElasticNet.pkl')
model_features = model.feature_names_in_
model_features
```

```python
import pickle

with open('drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/Tissue_res/y_train_features.pkl', 'rb') as file:
    y_features = pickle.load(file)

y_features = list(y_features)
```

```python
def merge_data(dictionary):

  for sample_rna, sample_met in dictionary.items():

    #Gene expression
    rna = pd.read_csv(f'drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/pancancer_metabolomics_v.0.3.4/pancancer_metabolomics/data/transcriptomics_processed/{sample_rna}', index_col = 0)
    rna.index.name = None
    rna = rna.T
    met_genes_select = met_genes[met_genes['Symbol'].isin(rna.columns)]
    met_genes_select = met_genes_select.Symbol.to_list()
    rna = rna[met_genes_select]
    rna = rna.T
    rna = pd.concat([met_genes.set_index('Symbol'), rna], axis = 'columns')
    rna = rna.set_index('gene_id')
    rna.index.name = None
    rna = rna.T
    rna = rna.reindex(model_features, axis = 'columns')


    met_id = sample_annot[sample_annot['RNAFile'].isin(microarrays)]
    met_id = met_id[['RNAID', 'MetabID']]
    met_id = met_id.set_index('RNAID')
    met_id.index.name = None
    rna = pd.concat([met_id, rna], axis = 'columns')
    rna = rna.set_index('MetabID')

    file_name = sample_rna.split('.')[0]
    rna.to_csv(f'drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/Microarrays/{file_name}_gene_expression.csv')
    print(str(file_name)+ '_gene_expression.csv' + ' saved!')

    #Metabolomics
    cohort_metabolomics = samples_dict[sample_rna]
    metabolomics = pd.read_excel(f'drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/pancancer_metabolomics_v.0.3.4/pancancer_metabolomics/data/metabolomics_processed/{cohort_metabolomics}', index_col = 0)
    common_metabolites = list(set(y_features).intersection(metabolomics.index))
    metabolomics = metabolomics.loc[common_metabolites]
    metabolomics = metabolomics.T

    metabolomics.to_csv(f'drive/My Drive/Colab Notebooks/Metabolite_prediction_ML/Microarrays/{file_name}_metabolomics.csv')
    print(str(file_name)+ '_metabolomics.csv' + ' saved!')
```

```python
merge_data(samples_dict)
```
